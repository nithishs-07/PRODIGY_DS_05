## Task 05: Traffic Accident Data Analysis & Pattern Recognition

**Objective**: Analyze traffic accident data to identify patterns related to road conditions, weather, and time of day. Visualize accident hotspots and contributing factors using the US Accidents dataset.

### 1. Import Libraries and Load Data

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)

# Load traffic accident sample dataset
df = pd.read_csv("US_Accidents_sample.csv")
print(f"Shape: {df.shape}")
df.head()
```

### 2. Timestamp Processing & Temporal Feature Extraction

```
df['Start_Time'] = pd.to_datetime(df['Start_Time'])
df['Hour'] = df['Start_Time'].dt.hour
df['Day_Of_Week'] = df['Start_Time'].dt.day_name()
df['Month'] = df['Start_Time'].dt.month_name()

df[['Start_Time', 'Hour', 'Day_Of_Week', 'Month']].head()
```

### 3. Temporal Patterns (Time of Day)

```
plt.figure(figsize=(10, 4))
hourly = df['Hour'].value_counts().sort_index()
sns.lineplot(x=hourly.index, y=hourly.values, marker='o', color='#e74c3c', linewidth=2.5)
plt.title('Traffic Accidents by Time of Day (24-Hour Cycle)', fontweight='bold')
plt.xlabel('Hour of Day (0-23)')
plt.ylabel('Accident Count')
plt.xticks(range(0, 24))
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### 4. Weather & Atmospheric Conditions

```
plt.figure(figsize=(8, 4))
top_weather = df['Weather_Condition'].value_counts().head(8)
sns.barplot(x=top_weather.values, y=top_weather.index, hue=top_weather.index, palette='Blues_r', legend=False)
plt.title('Accidents by Top Weather Conditions', fontweight='bold')
plt.xlabel('Accident Count')
plt.ylabel('Weather')
plt.show()
```

### 5. Road Infrastructure Hotspots

```
road_features = ['Traffic_Signal', 'Junction', 'Crossing', 'Station', 'Stop']
road_counts = {feat: df[feat].sum() for feat in road_features}
road_df = pd.DataFrame(list(road_counts.items()), columns=['Road_Feature', 'Count']).sort_values(by='Count', ascending=False)

plt.figure(figsize=(8, 4))
sns.barplot(data=road_df, x='Count', y='Road_Feature', hue='Road_Feature', palette='rocket', legend=False)
plt.title('Accident Frequency Near Road Infrastructure Elements', fontweight='bold')
plt.xlabel('Occurrences')
plt.ylabel('Feature')
plt.show()
```

### 6. Geographic Hotspots (Top Cities)

```
plt.figure(figsize=(8, 4))
top_cities = df['City'].value_counts().head(8)
sns.barplot(x=top_cities.values, y=top_cities.index, hue=top_cities.index, palette='magma', legend=False)
plt.title('Top Cities with Highest Accident Records', fontweight='bold')
plt.xlabel('Count')
plt.ylabel('City')
plt.show()
```

### 7. Severity Comparison: Day vs Night

```
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Severity', hue='Sunrise_Sunset', palette='Set2')
plt.title('Accident Severity Distribution: Day vs Night', fontweight='bold')
plt.xlabel('Severity Level (1 to 4)')
plt.ylabel('Count')
plt.legend(title='Time')
plt.show()
```

### 8. Key Findings
1. **Rush Hour Surges**: Accidents peak during the morning rush (7:00 AM – 9:00 AM) and evening commute (4:00 PM – 6:00 PM).
2. **Intersection Hotspots**: Locations with traffic signals and junctions account for the highest concentration of physical road incidents.
3. **Visibility & Lighting**: While most collisions occur during daytime owing to sheer vehicular traffic, night-time accidents present higher risk of severe delays due to reduced visibility.
